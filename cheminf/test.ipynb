```python
import pandas as pd  # "as pd" gives pandas a short nickname, the standard convention
from rdkit import Chem

df = pd.DataFrame({"name": ["ethanol", "benzene"], "smiles": ["CCO", "c1ccccc1"]})  # builds a small table from a dict: keys become column names, lists become the rows
df["mol"] = df["smiles"].apply(Chem.MolFromSmiles)  # .apply runs MolFromSmiles on every SMILES in the column and stores the results in a new column
df["atoms"] = df["mol"].apply(lambda m: m.GetNumAtoms())  # lambda is a one-line throwaway function; here it counts heavy atoms per molecule
print(df[["name", "atoms"]])  # double brackets select several columns and return a table
```
